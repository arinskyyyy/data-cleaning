### Sorting and filtering with Pokemon data

```python
#importing libraries
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("pokemon2.csv", skipinitialspace=True)
df.info()
df
```

#### 1. How many Pokemons exist with an Attack value greater than 150?

```python
df.loc[df['sp_attack'] > 150].shape

# another method
df.query('sp_attack > 150').shape
```

#### 2. Select all pokemons with a Speed of 10 or less

```python
df.loc[df['speed'] <= 10].shape
```

#### 3. How many Pokemons have a Sp. Def value of 25 or less?

```python
df.loc[df['sp_defense'] <= 25].shape
```

#### 4. Select all the Legendary pokemons

```python
legendary_df = df.loc[df['is_legendary'] == True].shape
legendary_df
```

#### 5. Find the outlier

```python
df.sort_values(by='sp_defense', ascending=False).head()

# sort by multiple criteria
df.sort_values(by=['sp_defense', 'speed'], ascending=[False, True]).head()
```

#### 6. How many Fire-Flying Pokemons are three?

```python
df.loc[(df['type1'] == 'fire') & (df['type2'] == 'flying')]
```

#### 7. How many 'Poison' pokemons are across both types?

```python
df.loc[(df['type1'] == 'poison') | (df['type2'] == 'poison')].shape
```

#### 8. What pokemon of Type 1 Ice has the strongest defense?

```python
df.loc[
    (df['type1'] == 'ice') &
    (df['sp_defense'] == df.loc[df['type1'] == 'ice', 'sp_defense'].max())
]
```

#### 9. What's the most common type of Legendary Pokemons?

```python
df.loc[df['is_legendary'] == 1, 'type1'].value_counts()
```

#### 10. What's the most powerful pokemon from the first 3 generations, of type water?

```python
df.loc[
    (df['type1'] == 'water') &
    (df['generation'].isin([1, 2, 3]))
].sort_values(by='generation', ascending=False).shape
```

#### 11. Select most powerful Fire-type pokemons

```python
df.loc[
    (df['type1'] == 'fire') & (df['sp_attack'] >= 100)
].sort_values(by='sp_attack', ascending=False).shape
```

#### 13. Select all Water-type, Flying-type pokemons

```python
df.loc[(df['type1'] == 'water') & (df['type2'] == 'flying')]
```

#### 14. Select specific columns of Legendary pokemons of type Fire

```python
df.loc[((df['type1'] == 'fire') | (df['type2'] == 'fire')) & (df['is_legendary'] == 1), ['sp_attack', 'generation']]
```

#### 15. Select Slow and Fast pokemons

```python
df['speed_rate'] = (df['speed'] / df['speed'].max()) * 100
df
```

```python
df.loc[(df['speed_rate'] >= 95) | (df['speed_rate'] <= 5)]
```

#### 16. Find the Ultra Powerful Legendary Pokemon

```python
df.loc[(df['is_legendary'] == 1)].sort_values(by=['sp_attack', 'sp_defense'], ascending=[False, False])
```
